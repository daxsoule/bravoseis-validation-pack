# Detection Validation — Phase A independent pass (reviewer: maleen)

Blind labeling pass. You will not see the other reviewer's labels. Output annotated PNGs go to `annotated_maleen/` and will be used to compute Cohen's kappa against the user's labels for inter-rater reliability.

## 4-class schema

- **TP** — real event, AIC pick is at the true first arrival
- **TP-peak** — real event, AIC pick lands on peak energy (late)
- **TP-coda** — real event, AIC pick lands in the coda / multipath (late)
- **FP** — spurious trigger, no real signal

## Pick adjustment

Each panel shows three vertical lines:

- **dashed blue** — raw STA/LTA pick
- **solid vermilion** — AIC-refined pick *(this is the picker output we are validating)*
- **dotted gray** — STA/LTA event end

**Hover** on the spectrogram or waveform to read the time in seconds. **Drag** to zoom for sub-second precision. Identify the visual true onset, type that time (in seconds within the panel) into `visual_onset_s` in the verdict cell. The render cell will compute `pick_shift_s = visual_onset_s − aic_pick_s` and draw a third **magenta** line at your onset on the saved annotated PNG.

If the event is **FP**, leave `visual_onset_s = None`.

## Workflow per event

1. Run the Plotly cell, hover/zoom to find the true onset.
2. Edit the `verdict = {...}` cell with `label`, `visual_onset_s`, `notes`.
3. Run the render cell — it writes the annotated PNG and a JSON sidecar.
4. Move on. The final cell aggregates all sidecars into a single CSV.

```python
import sys
from pathlib import Path
import pandas as pd
import plotly.io as pio

# Force inline rendering — works on JupyterHub without widget extensions.
pio.renderers.default = "notebook"

# Auto-detect the pack root (the directory containing pyproject.toml).
def _find_repo():
    for p in [Path.cwd()] + list(Path.cwd().parents):
        if (p / "pyproject.toml").exists():
            return p
    raise RuntimeError("Could not find repo root (looking for pyproject.toml)")

REPO = _find_repo()
sys.path.insert(0, str(REPO / "scripts"))
from validation_picker_panel import (
    load_event_data, make_plotly_panel, make_annotated_png,
)

REVIEWER = "maleen"
ANNOTATED_DIR = REPO / "outputs" / "figures" / "exploratory" / "validation" / f"annotated_{REVIEWER}"
ANNOTATED_DIR.mkdir(parents=True, exist_ok=True)

events = pd.read_csv(REPO / "outputs/figures/exploratory/validation/fp_validation_events.csv")
print(f"Reviewer: {REVIEWER}  |  events: {len(events)}  |  output: {ANNOTATED_DIR}")
```

## Event 1: E015371 — m6 / low / SNR 6.21 / grade A

```python
# --- Event 1: E015371 ---
d1 = load_event_data(1)
fig1 = make_plotly_panel(d1)
fig1.show()
```

```python
verdict1 = {
    'event_id': 'E015371',
    'idx': 1,
    'label': '',              # TP / TP-peak / TP-coda / FP
    'visual_onset_s': None,   # seconds within panel; None if FP
    'notes': '',              # free text
}
```

```python
# Render annotated PNG + JSON sidecar for event 1
import json as _json
from IPython.display import Image as _Image, display as _display
out_png = ANNOTATED_DIR / f'event_01_annotated.png'
out_json = ANNOTATED_DIR / f'event_01_verdict.json'
make_annotated_png(d1, verdict1, outpath=out_png, prior=None)
with open(out_json, 'w') as _f:
    _v = dict(verdict1)
    _v['reviewer'] = REVIEWER
    if _v.get('visual_onset_s') is not None:
        _v['pick_shift_s'] = float(_v['visual_onset_s']) - d1.aic_s
    else:
        _v['pick_shift_s'] = None
    _v['aic_pick_s'] = d1.aic_s
    _v['sta_lta_pick_s'] = d1.sta_lta_s
    _json.dump(_v, _f, indent=2)
print(f'  -> {out_png.name}')
_display(_Image(filename=str(out_png), width=1100))
```
