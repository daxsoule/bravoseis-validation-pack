# Phase 3 Catalogue Validation — independent pass (reviewer: maleen)

Blind labeling pass on a 25-seismic + 25-cryogenic catalogue-stratified sample. You will not see the primary reviewer's labels. Cohen's κ computed against the user pass for inter-rater reliability on the published-catalogue claim.

## 4-class schema (same as Phase A)

- **TP** — real event, AIC pick at the true first arrival (or within ~0.1 s)
- **TP-peak** — real event, AIC pick on peak energy (late)
- **TP-coda** — real event, AIC pick in the coda / multipath (late)
- **FP** — spurious trigger, no real signal

Picking convention: time-series SNR onset (PMEL — Fox 2001 / Dziak 2010), *provisional pending Bob's review*. Spectrogram first-energy times go in the `notes` field, not in `visual_onset_s`.

## Per-event workflow

1. Run the Plotly cell — hover/zoom to find the visual onset.
2. Edit the `verdict = {label, visual_onset_s, notes}` cell.
3. Run the render cell — writes the annotated PNG and a JSON sidecar.
4. After all 50, run the aggregator cell at the bottom for a single CSV.

```python
%load_ext autoreload
%autoreload 2

import sys
from pathlib import Path
import pandas as pd
import plotly.io as pio

pio.renderers.default = "notebook"

def _find_repo():
    for p in [Path.cwd()] + list(Path.cwd().parents):
        if (p / "pyproject.toml").exists():
            return p
    raise RuntimeError("Could not find repo root (looking for pyproject.toml)")

REPO = _find_repo()
sys.path.insert(0, str(REPO / "scripts"))
from validation_picker_panel import (
    load_event_data, make_plotly_panel, make_annotated_png,
)

REVIEWER = "maleen"
SAMPLE = "phase3"
SAMPLE_CSV = "fp_validation_events_phase3.csv"

ANNOTATED_DIR = REPO / "outputs" / "figures" / "exploratory" / "validation" / f"annotated_{SAMPLE}_{REVIEWER}"
ANNOTATED_DIR.mkdir(parents=True, exist_ok=True)

events = pd.read_csv(REPO / "outputs/figures/exploratory/validation" / SAMPLE_CSV)
print(f"Sample: {SAMPLE}  |  reviewer: {REVIEWER}  |  events: {len(events)}  |  output: {ANNOTATED_DIR}")
print(f"Class breakdown: {events["phase3_class"].value_counts().to_dict()}")
```

## Event 1: E007776 — m3 / seismic (low band) / SNR 10.71 / grade A

```python
# --- Event 1: E007776 (seismic) ---
d1 = load_event_data(1, csv_name=SAMPLE_CSV)
fig1 = make_plotly_panel(d1)
fig1.show()
```

```python
verdict1 = {
    'event_id': 'E007776',
    'idx': 1,
    'phase3_class': 'seismic',
    'label': '',              # TP / TP-peak / TP-coda / FP
    'visual_onset_s': None,   # seconds within panel; None if FP
    'notes': '',              # free text
}
```

```python
# Render annotated PNG + JSON sidecar for event 1
import json as _json
from IPython.display import Image as _Image, display as _display
out_png = ANNOTATED_DIR / f'event_01_annotated.png'
out_json = ANNOTATED_DIR / f'event_01_verdict.json'
make_annotated_png(d1, verdict1, outpath=out_png, prior=None)
with open(out_json, 'w') as _f:
    _v = dict(verdict1)
    _v['reviewer'] = REVIEWER
    _v['sample'] = SAMPLE
    if _v.get('visual_onset_s') is not None:
        _v['pick_shift_s'] = float(_v['visual_onset_s']) - d1.aic_s
    else:
        _v['pick_shift_s'] = None
    _v['aic_pick_s'] = d1.aic_s
    _v['sta_lta_pick_s'] = d1.sta_lta_s
    _json.dump(_v, _f, indent=2)
print(f'  -> {out_png.name}')
_display(_Image(filename=str(out_png), width=1100))
```
